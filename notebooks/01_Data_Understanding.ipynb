### Column Explanation and Grouping

objective:
- to organise the dataset into logical groups based on the information each column represents
- To understand the analytical role of each column group and how it will support answering the project's business questions

##### TIME DIMENSION

column(s): 
 - year_month

- this column will allow monthly comparison and time based analysis
- column will be split into two columns: year and month to enable analysis

##### GEOGRAPHY/NHS ORGANIZATION

    columns: 
     - region_name
    
     - region_code
    
     - icb_name
    
     - icb_code

- these columns will allow geographical analysis (i.e. regional comparison, board level comparison)
- will allow identification of regional prescription variations

##### DRUG IDENTIFICATION

    columns: 
     - bnf_presentation_name
    
     - bnf_presentation_code
    
     - snomed_code (clinical code)

- these columns will allow the analysis of medicines being prescribed
- will enable comparison of prescribing activity across products

##### DRUG CLASSIFICATION / PRODUCT HIERARCHY

    columns: 
     - generic_bnf_equivalent_name
    
     - generic_bnf_equivalent_code
    
     - bnf_chemical_substance_code
    
     - bnf_chemical_substance

- these columns group the medicines at the clinical and product level
- they can help in generic vs proprietary analysis
- they can enable active-substance level prescribing analysis

##### THERAPEUTIC CLASSIFICATION

    columns: 
    - bnf_chapter_code
        
    - bnf_chapter
        
    - bnf_section_code
        
    - bnf_section
         
    - bnf_paragraph_code
         
    - bnf_paragraph

- Enables analysis of prescribing burden by therapeutic area and disease group
- Supports identification of the main clinical domains driving prescribing activity and expenditure
- Allows comparison of resource utilisation across different therapeutic classes
- Provides a structured hierarchy for aggregating medicines at multiple levels of clinical detail

Insight Potential

- Identifies which disease areas contribute most to NHS prescribing activity
- Highlights high-cost therapeutic areas and their relative impact on overall expenditure
- Enables drill-down analysis from broad clinical categories (e.g. cardiovascular) to specific drug classes

##### Prescribing Activity and Cost

    columns: 
     - items (number of prescription items; measures drug utilisation) 
    
     - total_quantity (quantity of the medicine dispensed; measures volume)
    
     - nic (net ingredient cost; measures expenditure on medication)

- these columns will allow cost, volume and utilisation analysis
- will allow identification of high-volume and high-cost medications, appliances and medical devices

##### Product Characteristics

    columns: 
     - prep_class (indicates the type of product dispensed, such as generic medicines, proprietary medicines, appliances, dressings, or medical devices.)
    
     - prescribed_prep_class (indicates the type of product originally prescribed)
    
     - supplier_name


- these columns support analysis of prescribing patterns and product categories
- can be used to assess generic versus proprietary prescribing behaviour
- these columns can provide insight into prescribing behaviour and product type
- they will allow analyses of generic verses proprietary prescribing patterns
- they can also be used to examine the contribution of medicines, appliances and dressings to overall prescribing activity and expenditure

# SUMMARY

#### Dataset Grain
Each row represents monthly prescribing activity for a specific medicine presentation, including prescribing volume, quantity dispensed, expenditure, and its classification within the BNF therapeutic hierarchy.

#### Potential Analytical Opportunities

Initial exploration of the dataset suggests it can support the following areas of analysis:

- ##### Prescribing volume and utilisation patterns

    Understanding which medicines are most frequently prescribed and how prescribing activity varies across products and time.
  
- ##### Expenditure drivers

    Identifying the medicines and therapeutic areas that contribute most significantly to overall NHS prescribing costs.

- ##### Therapeutic burden across BNF chapters

    Analysing prescribing distribution across major disease areas to highlight key healthcare priorities.

- ##### Product type comparisons

    Comparing generic vs proprietary medicines, as well as distinguishing between medicines, appliances, and devices.

  
- ##### Temporal prescribing trends

    Exploring how prescribing activity and expenditure change over time (monthly trends).

These exploratory directions form the foundation for the subsequent data quality assessment and exploratory data analysis phases of the project.
